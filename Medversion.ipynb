```python
import pandas as pd
from neo4j import GraphDatabase
import os

# Your Neo4j connection details
NEO4J_URI = "neo4j+s://37631ae9.databases.neo4j.io"
NEO4J_USERNAME = "neo4j"
NEO4J_PASSWORD = "Lp0eU6JGKX_JW3nmy8hIT7sTtRcZSvvG-D4UGnqPsYY"

# File path to your CSV
CSV_PATH = "/content/drive/MyDrive/RESEARCH PROJECT/rwanda_health_centres.csv"

# Connect to Neo4j
driver = GraphDatabase.driver(NEO4J_URI, auth=(NEO4J_USERNAME, NEO4J_PASSWORD))

def clear_database():
    """Clear the database before loading new data"""
    print("Clearing existing data from the database...")

    with driver.session() as session:
        # Delete all nodes and relationships
        session.run("MATCH (n) DETACH DELETE n")
        print("Database cleared.")

def load_health_centers_with_relationships(csv_path):
    """
    Load health centers from CSV and create a proper graph model with relationships
    """
    print(f"Loading data from {csv_path}...")

    # Read the CSV file
    health_centers_df = pd.read_csv(csv_path)

    # Print summary of the data
    print(f"\nLoaded {len(health_centers_df)} health centers from CSV")
    print(f"Column names: {health_centers_df.columns.tolist()}")

    # Create Cypher query that builds a proper graph model
    query = """
    UNWIND $centers AS center

    // Create Admin1 node (province/region)
    MERGE (a:Admin1 {name: center.Admin1})

    // Create FacilityType node
    MERGE (ft:FacilityType {name: center.Facility_t})

    // Create Ownership node
    MERGE (o:Ownership {name: center.Ownership})

    // Create Country node
    MERGE (c:Country {name: center.Country})

    // Create HealthCenter node
    CREATE (h:HealthCenter {
        name: center.Facility_n,
        latitude: center.Lat,
        longitude: center.Long,
        ll_source: center.LL_source
    })

    // Create relationships
    MERGE (h)-[:LOCATED_IN]->(a)
    MERGE (h)-[:OF_TYPE]->(ft)
    MERGE (h)-[:OWNED_BY]->(o)
    MERGE (h)-[:IN_COUNTRY]->(c)
    """

    # Convert DataFrame to list of dictionaries
    centers_data = health_centers_df.to_dict('records')

    # Execute the query to create the nodes and relationships
    with driver.session() as session:
        session.run(query, centers=centers_data)

    # Verify the data was created correctly
    with driver.session() as session:
        # Count health centers
        result = session.run("MATCH (h:HealthCenter) RETURN COUNT(h) AS count")
        health_center_count = result.single()["count"]

        # Count Admin1 regions
        result = session.run("MATCH (a:Admin1) RETURN COUNT(a) AS count")
        admin1_count = result.single()["count"]

        # Count facility types
        result = session.run("MATCH (ft:FacilityType) RETURN COUNT(ft) AS count")
        facility_type_count = result.single()["count"]

        # Count ownership types
        result = session.run("MATCH (o:Ownership) RETURN COUNT(o) AS count")
        ownership_count = result.single()["count"]

        print(f"\nCreated {health_center_count} HealthCenter nodes")
        print(f"Created {admin1_count} Admin1 (region) nodes")
        print(f"Created {facility_type_count} FacilityType nodes")
        print(f"Created {ownership_count} Ownership nodes")

def display_graph_schema():
    """Display the current schema of the graph database"""
    print("\nCurrent Graph Schema:")

    with driver.session() as session:
        # Get node labels
        result = session.run("CALL db.labels()")
        labels = [record["label"] for record in result]
        print(f"Node Labels: {labels}")

        # Get relationship types
        result = session.run("CALL db.relationshipTypes()")
        rel_types = [record["relationshipType"] for record in result]
        print(f"Relationship Types: {rel_types}")

        # For each node label, get properties
        print("\nNode Properties:")
        for label in labels:
            result = session.run(
                f"MATCH (n:{label}) RETURN DISTINCT keys(n) AS properties LIMIT 1"
            )
            properties = result.single()["properties"] if result.peek() else []
            print(f"  {label}: {properties}")

def run_sample_queries():
    """Run some sample Cypher queries to test the data"""
    print("\nRunning sample queries:")

    queries = [
        {
            "description": "Count health centers by region",
            "query": """
            MATCH (h:HealthCenter)-[:LOCATED_IN]->(a:Admin1)
            RETURN a.name AS Region, COUNT(h) AS Count
            ORDER BY Count DESC
            """
        },
        {
            "description": "Count health centers by facility type",
            "query": """
            MATCH (h:HealthCenter)-[:OF_TYPE]->(ft:FacilityType)
            RETURN ft.name AS FacilityType, COUNT(h) AS Count
            ORDER BY Count DESC
            """
        },
        {
            "description": "Count health centers by ownership",
            "query": """
            MATCH (h:HealthCenter)-[:OWNED_BY]->(o:Ownership)
            RETURN o.name AS Ownership, COUNT(h) AS Count
            ORDER BY Count DESC
            """
        },
        {
            "description": "List all regions",
            "query": """
            MATCH (a:Admin1)
            RETURN a.name AS Region
            ORDER BY Region
            """
        }
    ]

    with driver.session() as session:
        for query_item in queries:
            print(f"\n{query_item['description']}:")
            result = session.run(query_item['query'])

            # Print results in a tabular format
            if result.peek():
                keys = result.keys()
                print(" | ".join(keys))
                print("-" * (sum(len(k) for k in keys) + 3 * (len(keys) - 1)))

                for record in result:
                    values = [str(record[key]) for key in keys]
                    print(" | ".join(values))
            else:
                print("No results found.")

# Main execution
if __name__ == "__main__":
    try:
        # Verify file exists
        if not os.path.exists(CSV_PATH):
            print(f"Error: File not found: {CSV_PATH}")
            exit(1)

        # Clear existing data
        clear_database()

        # Load the health centers with relationships
        load_health_centers_with_relationships(CSV_PATH)

        # Display the graph schema
        display_graph_schema()

        # Run sample queries
        run_sample_queries()

        print("\nData loading and modeling completed successfully!")
    except Exception as e:
        print(f"Error: {str(e)}")
    finally:
        # Always close the driver
        driver.close()
```

```python
# Install required packages
!pip install neo4j langchain-groq langchain

import os
from neo4j import GraphDatabase
from langchain_groq import ChatGroq
from langchain.schema import HumanMessage, SystemMessage

# Neo4j connection details
NEO4J_URI = "neo4j+s://37631ae9.databases.neo4j.io"
NEO4J_USERNAME = "neo4j"
NEO4J_PASSWORD = "Lp0eU6JGKX_JW3nmy8hIT7sTtRcZSvvG-D4UGnqPsYY"

# Updated Groq API Key
GROQ_API_KEY = "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"

class RwandaHealthcareSystem:
    def __init__(self):
        # Connect to Neo4j
        self.driver = GraphDatabase.driver(NEO4J_URI, auth=(NEO4J_USERNAME, NEO4J_PASSWORD))

        # Connect to Groq
        self.llm = ChatGroq(groq_api_key=GROQ_API_KEY, model_name="llama3-8b-8192")

        # Get database schema information
        self.schema_info = self._get_schema_info()
        print(f"Schema information collected. Ready to answer questions.")

    def _get_schema_info(self):
        """Get information about the database schema directly from the database"""
        schema_info = {}

        # Get node labels
        labels = self.run_query("CALL db.labels()")
        schema_info["node_labels"] = [record.get("label") for record in labels]

        # Get relationship types
        rel_types = self.run_query("CALL db.relationshipTypes()")
        schema_info["relationship_types"] = [record.get("relationshipType") for record in rel_types]

        # Get counts for main node types
        for label in ["Doctor", "HealthCenter", "Specialty"]:
            count = self.run_query(f"MATCH (n:{label}) RETURN count(n) as count")[0]["count"]
            schema_info[f"{label.lower()}_count"] = count

        # Sample some node properties for each label
        schema_info["node_properties"] = {}
        for label in schema_info["node_labels"]:
            props = self.run_query(f"MATCH (n:{label}) RETURN n LIMIT 1")
            if props:
                props_dict = dict(props[0]["n"])
                schema_info["node_properties"][label] = list(props_dict.keys())

        # Format as a readable string
        result = "DATABASE SCHEMA INFORMATION:\n\n"
        result += f"Node types: {', '.join(schema_info['node_labels'])}\n"
        result += f"Relationship types: {', '.join(schema_info['relationship_types'])}\n\n"
        result += f"Counts:\n"
        result += f"- {schema_info.get('doctor_count', 0)} Doctors\n"
        result += f"- {schema_info.get('healthcenter_count', 0)} Health Centers\n"
        result += f"- {schema_info.get('specialty_count', 0)} Specialties\n\n"

        result += "Node Properties:\n"
        for label, props in schema_info.get("node_properties", {}).items():
            result += f"- {label}: {', '.join(props)}\n"

        # Get sample data
        result += "\nSAMPLE DATA INSIGHTS:\n"

        # Sample hospitals
        hospitals = self.run_query("""
            MATCH (h:HealthCenter)
            WHERE h.name CONTAINS 'Military' OR h.name CONTAINS 'King Faisal'
            RETURN h.name
        """)
        hospital_names = [h["h.name"] for h in hospitals]
        result += f"Major hospitals: {', '.join(hospital_names)}\n"

        # Sample specialties
        specialties = self.run_query("MATCH (s:Specialty) RETURN s.name LIMIT 10")
        specialty_names = [s["s.name"] for s in specialties]
        result += f"Specialties include: {', '.join(specialty_names)}\n"

        # Sample admin regions
        regions = self.run_query("MATCH (a:Admin1) RETURN a.name LIMIT 10")
        region_names = [r["a.name"] for r in regions]
        result += f"Administrative regions include: {', '.join(region_names)}\n"

        # Sample doctor query
        doctors = self.run_query("""
            MATCH (d:Doctor)-[:WORKS_AT]->(h:HealthCenter)
            WHERE h.name CONTAINS 'Military'
            RETURN d.name, h.name
            LIMIT 3
        """)
        if doctors:
            result += f"\nExample: Doctors at Rwanda Military Hospital include: {', '.join([d['d.name'] for d in doctors])}\n"

        return result

    def close(self):
        """Close the Neo4j connection"""
        self.driver.close()

    def run_query(self, query, params=None):
        """Run a Cypher query and return the results"""
        with self.driver.session() as session:
            try:
                result = session.run(query, params or {})
                return [dict(record) for record in result]
            except Exception as e:
                print(f"Query error: {str(e)}")
                return []

    def generate_cypher_query(self, natural_language_query):
        """Use Groq to generate a Cypher query from a natural language question"""
        system_message = f"""You are an expert in converting natural language questions about healthcare in Rwanda into Cypher queries for a Neo4j graph database.

Here is information about the database schema:

{self.schema_info}

Your task is to create a correct Cypher query that will answer the user's question.

Rules for generating Cypher queries:
1. Use the exact node labels and relationship types shown above
2. For string matching, use CONTAINS for flexibility
3. When finding information about doctors, include their specialty and hospital information
4. Don't use properties that don't exist in the schema
5. Ensure all relationship directions are correct: (Doctor)-[:WORKS_AT]->(HealthCenter) not the other way around
6. Return meaningful column names using AS
7. Limit results to 20 if returning many rows
8. Return ONLY the Cypher query, nothing else"""

        messages = [
            {"role": "system", "content": system_message},
            {"role": "user", "content": f"Question: {natural_language_query}\n\nGenerate a Cypher query to answer this question."}
        ]

        response = self.llm.invoke(messages)
        return response.content

    def format_results(self, query_result):
        """Format the query results into a readable string"""
        if not query_result:
            return "No results found."

        # Get the keys from the first result
        keys = query_result[0].keys()

        # Calculate column widths (minimum 15, maximum 40)
        col_widths = {k: min(40, max(15, len(str(k)), max([len(str(r.get(k, ""))) for r in query_result]))) for k in keys}

        # Create header row
        header = " | ".join([f"{k:{col_widths[k]}}" for k in keys])
        separator = "-" * (sum(col_widths.values()) + 3 * (len(keys) - 1))

        # Create data rows
        rows = []
        for record in query_result:
            row = " | ".join([f"{str(record.get(k, '')):{col_widths[k]}}" for k in keys])
            rows.append(row)

        # Combine all parts
        return f"{header}\n{separator}\n" + "\n".join(rows)

    def explain_results(self, question, query, results):
        """Use Groq to explain the results in natural language"""
        # Format results to a string if they're not already
        if isinstance(results, list):
            formatted_results = self.format_results(results)
        else:
            formatted_results = str(results)

        # Handle case where results are too large
        if len(formatted_results) > 4000:
            formatted_results = formatted_results[:4000] + "...\n[Results truncated due to size]"

        system_message = """You are a helpful assistant explaining healthcare information about Rwanda.
Your task is to interpret query results from a healthcare database and provide a clear, helpful response to the user's question."""

        messages = [
            {"role": "system", "content": system_message},
            {"role": "user", "content": f"""Question: {question}

Cypher query used:
{query}

Query results:
{formatted_results}

Please provide a helpful, natural language answer to the question based on these results.
If there are no results, suggest possible reasons why or alternatives to try."""}
        ]

        response = self.llm.invoke(messages)
        return response.content

    def answer_question(self, question):
        """Process a natural language question about Rwanda healthcare"""
        print(f"Question: {question}")
        print("\nGenerating Cypher query...")

        # Generate Cypher query
        try:
            cypher_query = self.generate_cypher_query(question)
            print(f"Generated Cypher query: \n{cypher_query}\n")

            # Execute query
            print("Executing query...")
            results = self.run_query(cypher_query)

            # Format results for display
            formatted_results = self.format_results(results)
            print(f"Results:\n{formatted_results}\n")

            # Generate explanation
            print("Generating explanation...")
            explanation = self.explain_results(question, cypher_query, results)

            return {
                "question": question,
                "query": cypher_query,
                "raw_results": results,
                "formatted_results": formatted_results,
                "explanation": explanation
            }
        except Exception as e:
            print(f"Error: {str(e)}")
            return {
                "question": question,
                "error": str(e)
            }

def interactive_mode():
    """Run an interactive session with the Rwanda Healthcare System"""
    print("Initializing Rwanda Healthcare System...")
    system = RwandaHealthcareSystem()

    print("\n===== Rwanda Healthcare Query System =====")
    print("Ask questions about doctors, specialties, and healthcare facilities in Rwanda")
    print("Type 'exit' to quit. Type 'direct:' to run a direct Cypher query.")
    print("-" * 50)

    try:
        while True:
            user_input = input("\nYour question: ")

            if user_input.lower() in ['exit', 'quit', 'q']:
                break

            # Allow direct Cypher queries
            if user_input.startswith("direct:"):
                cypher_query = user_input[7:].strip()
                print(f"Running direct query: {cypher_query}")
                results = system.run_query(cypher_query)
                formatted_results = system.format_results(results)
                print(f"Results:\n{formatted_results}")
                continue

            # Process natural language question
            result = system.answer_question(user_input)

            if "error" in result:
                print(f"Error: {result['error']}")
            else:
                print("\nAnswer:")
                print(result["explanation"])

            print("-" * 50)
    finally:
        system.close()

# Run the interactive mode directly
if __name__ == "__main__":
    interactive_mode()
```
